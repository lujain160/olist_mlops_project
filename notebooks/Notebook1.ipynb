### Aggregate before you join — order items and payments have many rows per order

```python
# Items Aggregate
items_agg=items_table.groupby("order_id").agg(
    total_price=("price","sum"),
    avg_item_price=("price","mean"),
    max_item_price=("price","max"),
    min_item_price=("price","min"),
    total_freight=("freight_value","sum"),
    total_items=("order_item_id","count"),
    seller_id=("seller_id","first")
).reset_index()
items_agg.head()
```

```python
# Pyments Aggregate
pyments_agg=pyments_table.groupby("order_id").agg(
    total_pyment_value=("payment_value","sum"),
    max_installments=("payment_installments","max"),
    primary_pyment_type=("payment_type",lambda x: x.mode()[0] if not x.mode().empty else x.iloc[0])
).reset_index()
pyments_agg.head()
```

```python
# Reviews Aggregate
reviews_agg=reviews_table.groupby("order_id").agg(
    avg_review_score=("review_score","mean")
).reset_index()
reviews_agg.head()
```

### Keep the analysis light here: only what you need to join correctly

```python
ml_table=orders_table.merge(customers_table,on="customer_id",how="left") \
                     .merge(items_agg,on="order_id",how="left")\
                     .merge(pyments_agg,on="order_id",how="left")\
                     .merge(reviews_agg,on="order_id",how="left")
print(f"The final shape --> {ml_table.shape}")
ml_table.head()
```
